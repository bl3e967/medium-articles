# Creating My Own AI Agent into a Mobile App

## Why? 

A few months back, I wrote an article called 'Creating my own Voice Assistant using LangGraph and MCP servers' - where I leveraged the latest of LLMs to replicate and improve on the functionalities that existing home voice assistants provide.

Since then, I wanted to go a step further and extend the Voice Assistant from a service simple that runs on my local computer to a fully fledged mobile app.

It would be an opportunity to:
- Test out the latest LLM coding agents to develop a mobile app, something I've never done before.
- Learn about the whole developer/product lifecycle - from UI development, backend, platform and finally to deployment.

This article is to share the process of what I did and how I did it.

## Defining the mobile app I want to build

The project started off with creating my own Jarvis, a general assistant that could help me with anything.

I created my own MCP servers to slowly build out the capabilities of Jarvis:
- Adding the ability to connect with home devices, like light bulbs and my TV, and control them.
- Internet search, weather, time 
- Access to local file storage, access to Obsidian vaults, to create and manage documents.

But creating a general assistant soon turned out to be, frankly, boring. Why? I think because the scope was too broad and with no clearly defined end goal, psychologically it didn't feel intriguing anymore and my excitement was soon replaced with dread.

So I took a few weeks to think about what I could build that was narrower in scope, more achievable, and something related to something I genuinely am passionate about.

## Pivoting to a Cooking Assistant

I decided to marry my two passion together, Data Science and Cooking:

- Create a mobile app for my voice assistant that I can use for cooking.

The idea is, given a recipe, the voice assistant will guide me through the recipe step by step, so that I don't need to look at the recipe at all and everything becomes hands free.
What's more, the voice assistant would be able to dynamically change the recipe and steps depending on the ingredients or equipment I have.

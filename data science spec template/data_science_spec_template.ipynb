# The Data Scientist's Spec Template: How I Stopped Winging It and Started Shipping Better Code

*A practical, fill-in-the-blanks spec template for solo Python developers and data scientists — built from real projects, not enterprise theory.*

## Why?

Every article about spec-driven development (SDD) is written for backend engineers working in teams. There's a product manager, a tech lead, a Jira board, and a Git workflow with branch protection. The spec is a coordination tool.

But I'm a data scientist. I work solo, on personal projects, with no one to coordinate with except my future self and an AI coding assistant. The existing templates don't fit. They ask for stakeholder sign-off, API contract owners, and rollback plans. I just need to know what I'm building before I start building it.

So I wrote my own template. This article is that template — annotated, explained, and shown in use on a real project. If you're a data scientist or solo Python developer who's tired of vibe-coding your way into architectural dead ends, this is for you.

## Goals

### 1. Show what a data science spec actually looks like
Not a hypothetical. A real spec, for a real project, at multiple stages of refinement.

### 2. Give you a reusable template
Copy it, fill it in, adapt it. It should take you 30–60 minutes to complete — not a week.

### 3. Make the case for quality over speed
SDD isn't just faster. Done right, it produces *better* software — more coherent, fewer architectural U-turns, cleaner separation of concerns.

## 1. Why Data Scientists Need a Different Kind of Spec

The classic SDD stack assumes a team. The spec is a *communication* tool — it aligns engineers, product managers, and QA before anyone writes a line of code. That framing doesn't map cleanly to solo work.

When you're working alone, the spec serves a different purpose. It's a *thinking* tool. You're not aligning with other people — you're aligning with yourself. Forcing yourself to write down what you want to build exposes all the things you don't yet know about what you want to build.

> The spec doesn't just describe the thing you're going to build. It reveals the thing you actually want, underneath the thing you thought you wanted.

Data science projects have a few specific failure modes that a good spec can prevent:

- **Scope creep into experimentation.** You start building a pipeline, then decide to also try three different models, then realise you've lost track of what "done" looks like.
- **Unclear success criteria.** What does this model need to achieve to be useful? If you can't answer that before you start, you'll be answering it after weeks of work.
- **Interface ambiguity.** Who or what consumes the output? A dashboard? An API? Another script? Getting this wrong late is expensive.
- **Reproducibility as an afterthought.** Seeding, environment pinning, and data versioning are always "I'll add that later" — until they're not.

A well-written spec forces you to answer all of these upfront.

## 2. The Template

Here it is. Each section has a short explanation of *why* it's there, followed by the fill-in-the-blanks prompt.

---

### Section 1: Problem Statement

**Why:** This is the hardest section and the most important. If you can't write a clear problem statement, you're not ready to spec the solution. Most bad projects have a fuzzy problem statement at their root.

```
## Problem Statement

**What problem am I solving?**
<One paragraph. Be specific. Name the pain.>

**Who has this problem?**
<Me / my team / users of X / ...>

**What does the world look like if this is solved?**
<Concrete description of the end state. Not "faster" — faster how, by how much, for whom?>

**What's the current workaround, and why is it inadequate?**
<Be honest. If the workaround is fine, maybe you don't need this project.>
```

---

### Section 2: Success Criteria

**Why:** Data science projects rot without explicit success criteria. "Good accuracy" is not a criterion. "F1 score ≥ 0.85 on the held-out test set" is. This section is your exit condition — you'll know you're done when all of these are green.

```
## Success Criteria

**Must-have (the project fails without these):**
- [ ] <Criterion 1 — measurable>
- [ ] <Criterion 2 — measurable>

**Nice-to-have (stretch goals, explicitly labelled as optional):**
- [ ] <Criterion 3>

**Out of scope (explicitly excluded — prevents scope creep):**
- <Thing that sounds related but isn't part of this project>
```

---

### Section 3: Data

**Why:** Data is the dependency everything else hangs on. Spec it early. Discovering your data doesn't exist, isn't accessible, or has the wrong shape halfway through a project is a project-killer.

```
## Data

**Input data:**
- Source: <where does it come from?>
- Format: <CSV / parquet / SQL table / API response / ...>
- Size: <rough row count, file size, update frequency>
- Access: <do I have it now? If not, how do I get it?>
- Known issues: <missing values, label noise, class imbalance, ...>

**Output data:**
- Format: <what does the output look like?>
- Destination: <file, database, dashboard, API endpoint, ...>
- Consumer: <who or what reads this output?>

**Reproducibility requirements:**
- Random seed: <fixed? which?>
- Data versioning: <snapshot? DVC? just a timestamp?>
- Environment: <requirements.txt / pyproject.toml / conda env>
```

---

### Section 4: Architecture

**Why:** Even a small data science project has an architecture. Sketching it upfront — even badly — prevents the most painful refactors. This doesn't need to be a formal diagram. A bullet list of components and their relationships is enough.

```
## Architecture

**Components:**
1. <Component name> — <what it does, one sentence>
2. <Component name> — <what it does, one sentence>
3. ...

**Data flow:**
<Component A> → <Component B> → <Component C>

**Key interfaces:**
- <Component A> outputs: <format>
- <Component B> accepts: <format>, returns: <format>

**What I'm explicitly not building:**
- <Tempting extension that is out of scope>
```

---

### Section 5: Implementation Plan

**Why:** Break the work into phases, not tasks. Phases have a clear output — something runnable or testable at the end. Tasks are internal to a phase. This structure works well with AI-assisted development: you can hand a phase to an LLM with a clear contract for what it should produce.

```
## Implementation Plan

### Phase 1: <Name>
**Goal:** <What's the deliverable at the end of this phase?>
**Tasks:**
- [ ] Task 1
- [ ] Task 2
**Done when:** <How do I know this phase is complete?>

### Phase 2: <Name>
**Goal:** ...
**Tasks:** ...
**Done when:** ...

### Phase N: ...
```

---

### Section 6: Risks & Open Questions

**Why:** This is the section most people skip. Don't. The act of writing down your risks forces you to confront them while they're cheap to address. Open questions are things you don't know yet that could change the design — surface them early.

```
## Risks & Open Questions

**Known risks:**
- <Risk 1>: <Mitigation or contingency>
- <Risk 2>: <Mitigation or contingency>

**Open questions (must answer before starting Phase N):**
- [ ] <Question 1>
- [ ] <Question 2>

**Assumptions I'm making:**
- <Assumption 1>
- <Assumption 2>
```
